### 5 - Conteo de fuentes

```python
print(f"Cantidad de archivos CSV: {len(archivos_csv)}")
```

### 6 - Cargar tablas

```python
tablas = {}

for archivo in archivos_csv:
    nombre_tabla = archivo.stem
    tablas[nombre_tabla] = pd.read_csv(archivo)
```

### 7 - Inspeccion de datos

```python
# Tamaño de cada tabla
for nombre, df in tablas.items():
    print(f"{nombre}: {df.shape}")
```

### 8 Detalle de columnas


```python
detalle_columnas = []
for nombre_tabla, df in tablas.items():
    total_filas = len(df)
    
    for col in df.columns:
        conteo_nulos = df[col].isna().sum()
        pct_nulos = (conteo_nulos / total_filas * 100) if total_filas > 0 else 0.0
        
        detalle_columnas.append({
            "tabla": nombre_tabla,
            "columna": col,
            "tipo_dato": str(df[col].dtype),
            "no_nulos": df[col].notna().sum(),
            "nulos": conteo_nulos,
            "pct_nulos": round(pct_nulos, 2),
            "valores_unicos": df[col].nunique(dropna=True)
        })
```

```python
# Consolidar en un DataFrame
df_perfil = pd.DataFrame(detalle_columnas)
```

```python
# Vista previa de las primeras 15 columnas
print(f"Total de columnas analizadas en todo Northwind: {len(df_perfil)}")
print(df_perfil.head(15))
```

### 9 -  Deteccion de Fechas y Claves Primarias

```python
# 1. Identificar las columnas de fechas por nombre
columnas_fechas = df_perfil[df_perfil["columna"].str.contains("Date", case=False, na=False)]
print("=== COLUMNAS TEMPORALES (FECHAS) DETECTADAS ===")
display(columnas_fechas[["tabla", "columna", "tipo_dato", "no_nulos", "nulos"]])
# 2. Identificar candidatas a Clave Primaria (PK)
# Criterio: Sin nulos y valores únicos igual al total de filas de la tabla
candidatas_pk = []
for nombre_tabla, df in tablas.items():
    total_filas = len(df)
    for col in df.columns:
        if df[col].isna().sum() == 0 and df[col].nunique() == total_filas:
            candidatas_pk.append({
                "tabla": nombre_tabla,
                "columna_pk": col,
                "total_registros": total_filas,
                "valores_unicos": df[col].nunique()
            })
df_candidatas_pk = pd.DataFrame(candidatas_pk)
print("\n=== CANDIDATAS A CLAVE PRIMARIA (PK) ===")
display(df_candidatas_pk)
```

### 10 - Validacion de Clave compuesta en order-details

```python
df_od = tablas["order-details"]
total_filas_od = len(df_od)
# Conteo de combinaciones únicas entre orderID y productID
combinaciones_unicas = df_od.duplicated(subset=["orderID", "productID"]).sum()
print(f"Total registros en order-details: {total_filas_od}")
print(f"Duplicados en la combinación (orderID + productID): {combinaciones_unicas}")
if combinaciones_unicas == 0:
    print("[OK] Confirmado: (orderID + productID) es la Clave Primaria Compuesta válida.")
else:
    print("[ADVERTENCIA] Existen líneas duplicadas en el detalle de órdenes.")
```

### 11 - Conclusiones del EDA de Fuentes (Fase 1)

1. **Volumen de Fuentes:** 11 tablas analizadas con un total de 2,155 líneas transaccionales en `order-details` y 830 pedidos en `orders`.
2. **Tratamiento de Fechas:** Se identificaron 5 columnas temporales (`birthDate`, `hireDate`, `orderDate`, `requiredDate`, `shippedDate`) cargadas como texto (`str`), las cuales deberán castearse a `datetime` en la Capa Silver.
3. **Calidad y Nulos:**
   - `orders.shippedDate` cuenta con 21 nulos (2.53%), correspondientes a órdenes pendientes de despacho.
   - `employees.reportsTo` tiene 1 nulo, reflejando la cabeza jerárquica de la empresa.
   - `region` / `shipRegion` presentan >60% de nulos por variabilidad geográfica (se imputará "Sin Región" en Silver).
4. **Arquitectura de Claves:**
   - 10 tablas cuentan con claves primarias simples (`categoryID`, `customerID`, `orderID`, etc.).
   - La tabla transaccional `order-details` cuenta con una clave primaria compuesta verificada por `(orderID + productID)`.
